###Arithmetic example using different python data types.
-int
-float
-string
-Boolean
-Complex

```python
#integer 
a = 10
b = 5
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
#float
a = 10.5  
b= 2.5
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
#string
a = "10"  
b = "5"
print(a + b)  

a = "10"
b = "5"
print(int(a) + int(b))
```

```python
#boolean
a = True   
b = False
print(a + b)
print(a * b)
```

```python
#complex
a = 2 + 3j
b = 1 + 2j
print(a + b)
print(a - b)
print(a * b)
```

### comparision(relational)
>
<
>=
<=
==
!=

```python
#integer
a = 10
b = 5
print(a > b)
print(a < b)
print(a == b)
print(a != b)
```

```python
#float
a = 10.5
b = 5.5
print(a > b)
print(a < b)
print(a == b)
print(a != b)
```

```python
#string
a ="apple"
b = "banana"
print(a > b)
print(a < b)
print(a == b)
print(a != b)
```

```python
#boolean values can be compared.
#true is treated as ! and false as 0 for ordering comparision
a = True
b = False
print(a > b)
print(a == b)
print(a != b)
```

```python
#complex number do not support <,>,<=,or >== comparisions in python.
#you can only use == and !=.
a = 2 + 3j
b = 2 + 3j
print(a == b)
print(a != b)
```
